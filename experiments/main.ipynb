```python
def normalize_url(url: str) -> str:
    """Normalize the URL by removing fragments and queries."""
    parsed = urlparse(url)
    normalized = parsed._replace(fragment="", query="")
    return urlunparse(normalized).rstrip("/")
```

```python
normalized_url = normalize_url('https://www.geeksforgeeks.org/python/python-urllib-module/?query=example#section1')
print("Normalized URL:", normalized_url)
print("netloc:", urlparse(normalized_url).netloc)
```

```python
EXCLUDED_EXTENSIONS = [
    ".jpg",
    ".jpeg",
    ".png",
    ".gif",
    ".pdf",
    ".docx",
    ".xlsx",
    ".zip",
    ".rar",
    ".exe",
    ".svg",
    ".css",
    ".js",
]
```

```python
def is_valid_url(url: str) -> bool:
        """Validate the URL."""
        parsed = urlparse(url)
        if parsed.scheme not in ("http", "https") or not parsed.netloc:
            return False
        path = parsed.path.lower()
        return not any(path.endswith(ext) for ext in EXCLUDED_EXTENSIONS)
```

```python
valid = is_valid_url('https://www.geeksforgeeks.org/python/python-urllib-module/')
print("Is valid URL:", valid)
```

```python
import requests
from xml.etree import ElementTree
from typing import List, Set
from collections import deque
```

```python
def extract_domain(url: str) -> str:
        """Extract domain from a URL."""
        parsed_url = urlparse(url)
        return f"{parsed_url.scheme}://{parsed_url.netloc}"
```

```python
def crawl_sitemap(url: str) -> Set[str]:
        """Crawl the sitemap for all links, handling nested .xml sitemaps."""
        visited = set()
        all_urls = set()
        queue = deque([extract_domain(url) + "/sitemap.xml"])

        while queue:
            current_sitemap = queue.pop()
            if current_sitemap in visited:
                continue
            visited.add(current_sitemap)
            links = parse_sitemap(current_sitemap)
            for link in links:
                if link.endswith(".xml"):
                    queue.append(link)
                else:
                    all_urls.add(link)
        return all_urls
```

```python
results = crawl_sitemap("https://en.wikipedia.org/sitemap.xml")
print("Crawled URLs from sitemap:", results)
```

```python
def parse_sitemap(sitemap_url: str) -> List[str]:
        """Get URLs from sitemap."""
        try:
            response = requests.get(sitemap_url)
            response.raise_for_status()
            root = ElementTree.fromstring(response.content)
            namespace = {"ns": "http://www.sitemaps.org/schemas/sitemap/0.9"}
            return [loc.text for loc in root.findall(".//ns:loc", namespace)]
        except Exception as e:
            print(f"Error fetching sitemap: {e}")
            return []
```

```python
results = parse_sitemap("https://en.wikipedia.org/sitemap.xml")
print("Sitemap URLs:", results)
```

```python
class SitemapCrawler:
    def parse_sitemap(self, sitemap_url: str) -> List[str]:
        """Get URLs from sitemap."""
        try:
            response = requests.get(sitemap_url)
            response.raise_for_status()
            root = ElementTree.fromstring(response.content)
            namespace = {"ns": "http://www.sitemaps.org/schemas/sitemap/0.9"}
            return [loc.text for loc in root.findall(".//ns:loc", namespace)]
        except Exception as e:
            print(f"Error fetching sitemap: {e}")
            return []

    def crawl_sitemap(self, url: str) -> Set[str]:
        """Crawl the sitemap for all links, handling nested .xml sitemaps."""
        visited = set()
        all_urls = set()
        queue = deque([extract_domain(url) + "/sitemap.xml"])

        while queue:
            current_sitemap = queue.pop()
            if current_sitemap in visited:
                continue
            visited.add(current_sitemap)
            links = self.parse_sitemap(current_sitemap)
            for link in links:
                if link.endswith(".xml"):
                    queue.append(link)
                else:
                    all_urls.add(link)
        return all_urls
```
